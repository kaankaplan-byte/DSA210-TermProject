# Phase 3: Machine Learning Models (TMDb Popularity Prediction)

This notebook trains multiple regression models to predict **movie popularity** using the cleaned TMDb dataset
and reports evaluation metrics (**RMSE**, **MAE**, **R²**) on a test set.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/movies_clean.csv")
df.head()
```

```python
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
df.describe()
```

## Model Results (Test Set)

| Model | RMSE | MAE | R² |
|---|---:|---:|---:|
| RandomForest | 20.1257 | 10.2236 | -0.2175 |
| Ridge (baseline) | 20.2804 | 15.8989 | -0.2363 |
| GradientBoosting | 26.9913 | 10.9230 | -1.1898 |

## Interpretation

- **RandomForest** achieved the lowest RMSE among the tested models.
- However, all models have **negative R²**, meaning they do not outperform a simple baseline predictor
  that always predicts the mean popularity on the test set.
- This suggests that **popularity is driven by additional factors not captured by basic metadata**
  (e.g., marketing, social media trends, cast influence, release strategy, timing, and region).

### Possible Improvements
- Apply **log-transform** to highly skewed variables such as budget and revenue.
- Handle outliers (winsorization or filtering extreme values).
- Add richer TMDb signals: **keywords**, **cast/crew features**, production companies, language/region,
  and time-based features.
