```python
numeric_cols = ["budget","revenue","runtime","vote_average","vote_count","popularity"]
corr = df[numeric_cols].corr(method="spearman")
corr
```

```python
from scipy import stats

def spearman_test(x, y):
    d = df[[x, y]].dropna()
    r, p = stats.spearmanr(d[x], d[y])
    return r, p, len(d)

r, p, n = spearman_test("popularity", "budget")
print("TEST 1: Spearman(popularity, budget) -> r=", r, "p=", p, "n=", n)

r, p, n = spearman_test("popularity", "vote_average")
print("TEST 2: Spearman(popularity, vote_average) -> r=", r, "p=", p, "n=", n)
```

```python
def mann_whitney(a, b):
    a = a.dropna()
    b = b.dropna()
    u, p = stats.mannwhitneyu(a, b, alternative="two-sided")
    return u, p

if "Genre_Action" in df.columns:
    action = df[df["Genre_Action"] == 1]["popularity"]
    non_action = df[df["Genre_Action"] == 0]["popularity"]
    u, p = mann_whitney(action, non_action)
    print("TEST 3: Action vs Non-Action -> U=", u, "p=", p)
else:
    print("Genre_Action not found.")

df["release_date"] = pd.to_datetime(df["release_date"], errors="coerce")
pre = df[df["release_date"] < "2010-01-01"]["popularity"]
post = df[df["release_date"] >= "2010-01-01"]["popularity"]
u, p = mann_whitney(pre, post)
print("TEST 4: Pre-2010 vs 2010+ -> U=", u, "p=", p)
```

### Hypothesis Test Conclusions

Four hypothesis tests were conducted to analyze factors affecting movie popularity.

1. **Popularity vs Budget:**  
   A Spearman correlation test revealed a strong positive correlation (r = 0.612, p < 0.001),
   indicating that higher-budget movies tend to have higher popularity values. The null
   hypothesis was rejected.

2. **Popularity vs Vote Average:**  
   A statistically significant negative correlation was observed (r = -0.434, p < 0.001).
   This suggests that highly popular movies do not necessarily receive higher user ratings,
   potentially reflecting the difference between commercial success and critical reception.

3. **Genre Effect (Action vs Non-Action):**  
   A Mann–Whitney U test showed a significant difference in popularity distributions
   between Action and non-Action movies (p < 0.001), indicating that genre plays a
   meaningful role in popularity.

4. **Temporal Effect (Pre-2010 vs 2010+):**  
   Popularity distributions differed significantly between movies released before and
   after 2010 (p < 0.001). This may be attributed to changes in marketing strategies,
   digital platforms, and audience behavior in recent years.
